```python
def next_state(s, v, dt, c = 1, g = g):
    a = np.array([-c*v[0], -c*v[1], -c*v[2]-g])
    # print(f"{a.dtype = }")
    # print(f"{v.dtype = }")
    s_ = s + v*dt + 0.5*a*dt**2
    v_ = v + a*dt
    return s_, v_
```

```python
s0 = np.array([0.0, 0.0, 0.0])
v0 = np.array([1.0, 0.0, 1.0])
v0 /= np.linalg.norm(v0)/10
print(f"{v0 = }")
state_s = [s0]
state_v = [v0]
dt = 0.01
for i in range(int(2/dt)):
    s_, v_ = next_state(state_s[-1], state_v[-1], dt, c=1.5)
    state_s.append(s_)
    state_v.append(v_)
state_s = np.array(state_s)

x = state_s[:, 0]
y = state_s[:, 1]
z = state_s[:, 2]
plt.plot(x, z)
plt.xlim(0, 7)
plt.ylim(0, 7)
# aspect = 1
plt.gca().set_aspect('equal', adjustable='box')
```
